```python
#euler method

import matplotlib.pyplot as plt
dt = 0.1
g = -9.81
height = 2000
v = 0

times = []
heights = []

for x in range(200):
    v = v + (g*dt) 
    height = height + (v*dt)
    times.append(x*dt)
    heights.append(height)
    

plt.plot(times, heights)
plt.xlabel("Times (s)")
plt.ylabel("Height (m)")
plt.title("Falling ball euler method")
```

```python
import matplotlib.pyplot as plt
import numpy as np

g=-9.81
angle=np.radians(30)
v=60
height=100
dt=0.1

vx=v*np.cos(angle)
vy=v*np.sin(angle)

xposition = []
yposition = []

y=height
x=0

while y > 0:
    vx = vx + 0*dt
    vy = vy + g*dt
    x = x + vx*dt
    y = y + vy*dt 

    xposition.append(x)
    yposition.append(y)

plt.plot(xposition,yposition)
plt.xlabel("Horizontal distance (m)")
plt.ylabel("Vertical distance (m)")
plt.title("Projectile motion")
```

```python
import matplotlib.pyplot as plt
import numpy as np

g=-9.81
angle=np.radians(60)
v=60
height=600
dt=1

vx=v*np.cos(angle)
vy=v*np.sin(angle)

xposition = []
yposition = []

y=height
x=0


for step in range(100):
    vx = vx + 0*dt
    vy = vy + g*dt
    x = x + vx*dt
    y = y + vy*dt 
    
    if y<=0:
        vy = -vy*0.9
        vx = vx*0.99
        y=0
        

    xposition.append(x)
    yposition.append(y)

plt.plot(xposition,yposition)
plt.xlabel("Horizontal distance (m)")
plt.ylabel("Vertical distance (m)")
plt.title("Bouncy Ball")
```

```python

import matplotlib.pyplot as plt
import numpy as np

theta = np.radians(30) 
omega = 0
l = 20
g = -9.81
dt = 0.02
time = 0
total_time = 10

times = []
angles = []

for step in range(int(total_time / dt)):
    alpha = (g / l) * np.sin(theta)
    
    omega = omega + alpha * dt
    theta = theta + omega * dt
    time += dt

    times.append(time)
    angles.append(np.degrees(theta))

plt.plot(times, angles)
plt.xlabel("Time (s)")
plt.ylabel("Angle (degrees)")
plt.title("Simple Pendulum")
plt.grid()
plt.show()
```
