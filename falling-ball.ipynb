```python
#euler method

import matplotlib.pyplot as plt
dt = 0.1
g = -9.81
height = 2000
v = 0

times = []
heights = []

for x in range(200):
    v = v + (g*dt) 
    height = height + (v*dt)
    times.append(x*dt)
    heights.append(height)
    

plt.plot(times, heights)
plt.xlabel("Times (s)")
plt.ylabel("Height (m)")
plt.title("Falling ball euler method")
```

```python
import matplotlib.pyplot as plt
import numpy as np

g=-9.81
angle=np.radians(30)
v=60
height=100
dt=0.1

vx=v*np.cos(angle)
vy=v*np.sin(angle)

xposition = []
yposition = []

y=height
x=0

while y > 0:
    vx = vx + 0*dt
    vy = vy + g*dt
    x = x + vx*dt
    y = y + vy*dt 

    xposition.append(x)
    yposition.append(y)

plt.plot(xposition,yposition)
plt.xlabel("Horizontal distance (m)")
plt.ylabel("Vertical distance (m)")
plt.title("Projectile motion")
```
